```python
from meridian.analysis import optimizer


# Example 1: target total ROI
opt_flex_roi = budget_optimizer.optimize(
    fixed_budget=False,   # use FlexibleBudgetScenario
    target_roi=50.0,      # stop when total ROI reaches ~50
    use_kpi=True,
)

# Example 2: target marginal ROI
opt_flex_mroi = budget_optimizer.optimize(
    fixed_budget=False,
    target_mroi=1.0,      # stop when marginal ROI drops to ~1
    use_kpi=True,
)
```

```python
flex_nonopt = opt_flex_roi.nonoptimized_data
flex_opt = opt_flex_roi.optimized_data

flex_nonopt_df = flex_nonopt.sel(metric="mean").to_dataframe().reset_index()
flex_opt_df   = flex_opt.sel(metric="mean").to_dataframe().reset_index()

# 1) Rename columns to indicate scenario
flex_nonopt_df = flex_nonopt_df.rename(
    columns={
        "spend": "nonopt_spend",
        "incremental_outcome": "nonopt_incremental_outcome",
        "roi": "nonopt_roi",
        "mroi": "nonopt_mroi",
        "cpik": "nonopt_cpik",
    }
)
flex_opt_df = flex_opt_df.rename(
    columns={
        "spend": "opt_spend",
        "incremental_outcome": "opt_incremental_outcome",
        "roi": "opt_roi",
        "mroi": "opt_mroi",
        "cpik": "opt_cpik",
    }
)

# 2) Merge on channel (metric is the same 'mean' in both)
flex_compare_df = (
    flex_nonopt_df
    .drop(columns=["metric"])
    .merge(
        flex_opt_df.drop(columns=["metric"]),
        on="channel",
    )
)

# 3) Optional: pretty formatting
import pandas as pd

formatted_flex_df = flex_compare_df.copy()
for col in [
    "nonopt_spend", "opt_spend",
    "nonopt_incremental_outcome", "opt_incremental_outcome",
]:
    formatted_flex_df[col] = formatted_flex_df[col].map(lambda x: f"{x:,.0f}")
for col in [
    "nonopt_roi", "opt_roi",
    "nonopt_mroi", "opt_mroi",
    "nonopt_cpik", "opt_cpik",
]:
    formatted_flex_df[col] = formatted_flex_df[col].map(lambda x: f"{x:,.2f}")

formatted_flex_df  # final flexible-budget table
```

```python
from meridian.analysis import optimizer

fixed_opt_results = budget_optimizer.optimize(
    fixed_budget=True,
    budget=250_000_000.0,
    use_kpi=True,
)

# 1) Spend ratio (vs historical)
spend_ratio = fixed_opt_results.spend_ratio

# 2) Non-optimized, non-optimized with optimal freq, and optimized datasets
fixed_nonopt_ds = fixed_opt_results.nonoptimized_data
fixed_nonopt_optfreq_ds = fixed_opt_results.nonoptimized_data_with_optimal_freq
fixed_opt_ds = fixed_opt_results.optimized_data

# 3) Optimization grid (for response curves etc.)
fixed_opt_grid = fixed_opt_results.optimization_grid  # OptimizationGrid

# xr.Dataset with spend_grid, incremental_outcome_grid
fixed_grid_ds = fixed_opt_grid.grid_dataset

print(fixed_grid_ds)
print(fixed_grid_ds.data_vars)
```

```python
# 1) Channel-level summary for cards / tables
channels = fixed_grid_ds['channel'].data.tolist()

# Spend
nonopt_spend_df = fixed_nonopt_ds.spend.to_dataframe().reset_index()
opt_spend_df    = fixed_opt_ds.spend.to_dataframe().reset_index()

# Incremental outcome (mean metric)
nonopt_inc_df = (
    fixed_nonopt_ds.incremental_outcome
    .sel(metric='mean')
    .to_dataframe()
    .reset_index()
)
opt_inc_df = (
    fixed_opt_ds.incremental_outcome
    .sel(metric='mean')
    .to_dataframe()
    .reset_index()
)

# Merge
summary = (
    nonopt_spend_df
    .merge(opt_spend_df, on='channel', suffixes=('_nonopt_spend', '_opt_spend'))
    .merge(nonopt_inc_df, on='channel')
    .merge(opt_inc_df, on='channel', suffixes=('_nonopt_inc', '_opt_inc'))
)

# Inspect current columns (for clarity)
# print(summary.columns)
# Expected: ['channel', 'spend_nonopt_spend', 'spend_opt_spend',
#            'incremental_outcome_nonopt_inc', 'metric_nonopt_inc',
#            'incremental_outcome_opt_inc', 'metric_opt_inc']

# Rename only the value columns, drop metric columns
summary = summary.rename(columns={
    'spend_nonopt_spend': 'nonopt_spend',
    'spend_opt_spend': 'opt_spend',
    'incremental_outcome_nonopt_inc': 'nonopt_inc',
    'incremental_outcome_opt_inc': 'opt_inc',
})

summary = summary[['channel', 'nonopt_spend', 'opt_spend', 'nonopt_inc', 'opt_inc']]

# Derived metrics
summary['delta_spend'] = summary['opt_spend'] - summary['nonopt_spend']
summary['delta_inc']   = summary['opt_inc']   - summary['nonopt_inc']

summary['nonopt_spend_pct'] = summary['nonopt_spend'] / summary['nonopt_spend'].sum()
summary['opt_spend_pct']    = summary['opt_spend']    / summary['opt_spend'].sum()

summary['nonopt_inc_pct'] = summary['nonopt_inc'] / summary['nonopt_inc'].sum()
summary['opt_inc_pct']    = summary['opt_inc']    / summary['opt_inc'].sum()

summary['inc_lift_pct'] = (summary['delta_inc'] / summary['nonopt_inc']).round(4)

summary_df = summary.sort_values('opt_spend', ascending=False).reset_index(drop=True)


summary_df
```

```python
# Budget allocation table

budget_alloc = summary_df[['channel', 'nonopt_spend_pct', 'opt_spend_pct']].copy()
budget_alloc.columns = ['channel', 'nonopt_spend_pct', 'opt_spend_pct']

# For HTML table, you might also want nicely formatted strings:
budget_alloc['nonopt_spend_pct_str'] = (budget_alloc['nonopt_spend_pct'] * 100).round(1).astype(str) + '%'
budget_alloc['opt_spend_pct_str']    = (budget_alloc['opt_spend_pct'] * 100).round(1).astype(str) + '%'

budget_alloc.to_csv("budget_allocation_for_html.csv", index=False)
budget_alloc
```

```python
# 3) Response curve data per channel

spend_grid = fixed_grid_ds['spend_grid']                # (grid_spend_index, channel)
inc_grid   = fixed_grid_ds['incremental_outcome_grid']  # (grid_spend_index, channel)
channels   = fixed_grid_ds['channel'].data.tolist()

rows = []
for ch in channels:
    x = spend_grid.sel(channel=ch).values
    y = inc_grid.sel(channel=ch).values
    for i in range(len(x)):
        rows.append({
            "channel": ch,
            "grid_spend": float(x[i]),
            "grid_incremental_outcome": float(y[i]),
        })

response_curves_df = pd.DataFrame(rows)

# Optional: add markers for baseline / optimized points on top of grid
nonopt_spend_series = fixed_nonopt_ds.spend.to_dataframe()['spend']
opt_spend_series    = fixed_opt_ds.spend.to_dataframe()['spend']
nonopt_inc_series   = fixed_nonopt_ds.incremental_outcome.sel(metric='mean').to_dataframe()['incremental_outcome']
opt_inc_series      = fixed_opt_ds.incremental_outcome.sel(metric='mean').to_dataframe()['incremental_outcome']

markers = []
for ch in channels:
    markers.append({
        "channel": ch,
        "point_type": "baseline",
        "spend": float(nonopt_spend_series[ch]),
        "incremental_outcome": float(nonopt_inc_series[ch]),
    })
    markers.append({
        "channel": ch,
        "point_type": "optimized",
        "spend": float(opt_spend_series[ch]),
        "incremental_outcome": float(opt_inc_series[ch]),
    })

response_markers_df = pd.DataFrame(markers)


```

```python
import pandas as pd

channels = fixed_grid_ds['channel'].data.tolist()

# Use mean metric for incremental outcome
nonopt_inc = (
    fixed_nonopt_ds.incremental_outcome
    .sel(metric='mean')
    .to_dataframe()['incremental_outcome']
)
opt_inc = (
    fixed_opt_ds.incremental_outcome
    .sel(metric='mean')
    .to_dataframe()['incremental_outcome']
)

nonopt_spend = fixed_nonopt_ds.spend.to_dataframe()['spend']
opt_spend    = fixed_opt_ds.spend.to_dataframe()['spend']

# Simple ROI estimate at the optimized point
channel_perf_rows = []
for ch in channels:
    spend = float(opt_spend[ch])
    inc   = float(opt_inc[ch])
    roi   = inc / spend if spend > 0 else 0.0

    channel_perf_rows.append({
        "channel": ch,
        "baseline_spend": float(nonopt_spend[ch]),
        "optimized_spend": float(opt_spend[ch]),
        "baseline_incremental": float(nonopt_inc[ch]),
        "optimized_incremental": float(opt_inc[ch]),
        "roi": roi,                       # used as ch.roi in your JSX
    })

channel_performance_df = pd.DataFrame(channel_perf_rows)
channel_performance_df.to_csv("channel_performance_for_scenario_planning.csv", index=False)
channel_performance_df
```

```python
import json
import re

def to_camel(s: str) -> str:
    s = re.sub(r'[^0-9a-zA-Z]+', ' ', s).strip()
    parts = s.split()
    return parts[0].lower() + ''.join(p.capitalize() for p in parts[1:])

channel_keys = {ch: to_camel(ch) for ch in channels}

# Initialize budget at optimized spend (or baseline_spend if you prefer)
budget_dict = {channel_keys[ch]: float(opt_spend[ch]) for ch in channels}

with open("channel_keys.json", "w") as f:
    json.dump(channel_keys, f, indent=2)

with open("initial_budget.json", "w") as f:
    json.dump(budget_dict, f, indent=2)

channel_keys, budget_dict
```
